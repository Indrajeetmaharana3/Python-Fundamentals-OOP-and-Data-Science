# Swap variable in python

```python
a = 5
b = 6
```

```python
a = b
b = a
```

```python
a,b = b,a
```

```python
print(a)
print(b)
```

```python
# we can't swap in this method so,
a1 = 7
b1 = 8
```

```python
temp = a1        # using 3rd variable to swap is possible
a1 = b1
b1= temp
```

```python
print(a1)
print(b1)
```

```python
a2 = 5
b2 = 3
```

```python
a2 = a2+b2  # using mathemathical operation
b2 = a2-b2
a2 = a2-b2
```

```python
print(a2)
print(b2)
```
