# Decimal Number system

```python
a = 10
b = 25
c = 100

print(a)
print(b)
print(c)
```

```python
x = 10            # decimal to binary
print(bin(x))
```

```python
print(oct(x))      # decimal to octal
```

```python
print(hex(x))      # decimal to hexadecimal
```

# Binary Number system

```python
x = 0b1010

print(x)
```

```python
x = int('1010',2)      # binary to decimal
print(x)
```

```python
print(int('101', 2))
print(int('1100', 2))
print(int('1111', 2))
```

# Octal Number system

```python
x = 0o12
print(x)
```

```python
x = 10               # octal to decimal
print(oct(x))
```

# HexaDecimal Number system

```python
x = 0xA
print(x)
```

```python
x = int('A', 16)       # hexadecimal to decimal
print(x)
```

```python
x = 10
print(hex(x))
```

# coversion

```python
print(int('11001', 2))        # binary -> decimal
```

```python
print(int('31', 8))          # octal -> decimal
```

```python
print(int('19', 16))       # hexadecimal to decimal
```

```python
x = 25
print(bin(x))
print(oct(x))
print(x)
print(hex(x))
```

```python
num = 10
print(bin(num))
print(oct(num))
print(num)
print(hex(num))
```
