```python
110.34
```

```python
int(110.34)  # float to int possible
```

```python
int(True)  # boolean to int possible
```

```python
int(1+2j)  # complex to int Not possible
```

```python
int('10')  # N.string to int possible
```

```python
int('ten')  # T.string to int Not possible
```

```python
float(110)  # int to float possible
```

```python
float(True)  # boolean to float possible
```

```python
float(False)
```

```python
float(1+2j)  # complex to float not possible
```

```python
float('10')  # N.string to float possible
```

```python
float('ten')  # T.string to float not possible
```
