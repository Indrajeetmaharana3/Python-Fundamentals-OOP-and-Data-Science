```python
# Example
x = [1, 2, 3, 4, 5]

sales = [10, 20, 15, 25, 30]
profit = [5, 12, 10, 18, 22]

plt.plot(
    x,
    sales,
    marker="o",
    linestyle="--",
    linewidth=2,
    label="Sales"
)

plt.plot(
    x,
    profit,
    marker="s",
    linestyle="-",
    linewidth=2,
    label="Profit"
)

plt.title("Sales and Profit Report")
plt.xlabel("Days")
plt.ylabel("Amount")

plt.xlim(0, 6)
plt.ylim(0, 35)

plt.grid(
    linestyle="--",
    alpha=0.5
)

plt.legend(loc="upper left")

plt.savefig(
    "sales_profit.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import matplotlib.pyplot as plt
products = ["A", "B", "C", "D"]
sales = [20, 35, 30, 45]

plt.bar(products, sales)
plt.show()
```

```python
# Add color in bar
plt.bar(products, sales, color="Green")
plt.show()
```

```python
plt.bar(products, 
        sales,
        color = ["red","blue","green","orange"])
plt.show()
```

```python
# add barwidth in bar
plt.bar(products, 
        sales,
        width = 0.5,
        color = ["red","blue","green","orange"])
plt.show()
```

```python
# add alpha in bar
plt.bar(products, 
        sales,
        width = 0.5,
        alpha = 0.6,
        color = ["red","blue","green","orange"])
plt.show()
```

```python
# add edgecolor in bar
plt.bar(products, 
        sales,
        width = 0.5,
        alpha = 0.6,
        color = ["red","blue","green","orange"],
        edgecolor = "black")
plt.show()
```

```python
# add border thickness in bar
plt.bar(products, 
        sales,
        width = 0.5,
        alpha = 0.6,
        color = ["red","blue","green","orange"],
        edgecolor = "black",
        linewidth = 2)
plt.show()
```

```python
# add values on top of bars
plt.bar(products, sales)
for i in range (len(sales)):
    plt.text(i, sales[i] + 1, sales[i], ha = "center")
plt.show()
```

```python
#Ex:
products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [40, 70, 50, 30]
plt.bar(
    products,
    sales,
    color=["blue", "green", "orange", "purple"],
    width=0.6,
    alpha=0.8,
    edgecolor="black",
    linewidth=1.5
)

for i in range(len(sales)):
    plt.text(
        i,
        sales[i] + 1,
        sales[i],
        ha="center"
    )

plt.title("Product Sales", fontsize=18, fontweight="bold")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```
